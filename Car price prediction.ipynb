```python
# Encoding datachecking categorical dataset
car_data_set.replace({'Fuel_Type':{'Petrol':0,'Diesel':1,'CNG':2}},inplace=True)
car_data_set.replace({'Seller_Type':{'Dealer':0,'Individual':1}},inplace= True)
car_data_set.replace({'Transmission':{'Manual':0,'Automatic':1}},inplace= True)
```

```python
car_data_set.head()
```

```python
# Splitting data tranning data set test data Target data set

x = car_data_set.drop(['Selling_Price','Car_Name'],axis=1)
y = car_data_set['Selling_Price']
```

```python
x
```

```python
y
```

```python
# Splitting Trannig and Test Data 

x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.1,random_state = 2)
```

<h4><b>MODEL TRAINNING</b></h4>

```python
# Linear Regression Model

lin_reg_model=LinearRegression()

```

```python
lin_reg_model.fit(x_train,y_train)
```

<h4><b>MODEL EVOLUTION</b></h4>

```python
# prediction of trainning data

trainning_data_prediction = lin_reg_model.predict(x_train)
```

```python
# R Square Errors

error_score = metrics.r2_score(y_train,trainning_data_prediction)
print('r Squred error :',error_score)
```

```python
# Visulization the actual prices and prediction prices

plt.scatter(y_train,trainning_data_prediction)
plt.xlabel('Actual Prices')
plt.ylabel('Prediction Prices')
plt.title('Actual Prices vs Prediction Prices')
plt.show()
```

```python
# Test data set prediction
test_data_prediction = lin_reg_model.predict(x_test)
```

```python
# Test data score

error_score = metrics.r2_score(y_test,test_data_prediction)
print('R squred error:',error_score)
```

```python
# Testing data prediction 
plt.scatter(y_test,test_data_prediction)
plt.xlabel('Actual Prices')
plt.ylabel('Prediction Prices')
plt.title('Actual Prices vs Prediction Prices')
plt.show()
```
